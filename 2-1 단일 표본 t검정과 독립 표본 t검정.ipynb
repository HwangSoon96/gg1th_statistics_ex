# 단일 표본 t 검정

1. 표본 데이터 로딩
2. 표본 데이터가 정규성을 따르는지 검정
3. 단일 표본 t 검정
    - p-value < 0.05 : 기무가설 기각, 대립가설 채택, 통계적으로 유의미하다고 해석할 수 있음
    - p-value >= 0.05 : 기무가설 채택, 대립가설 기각


## 정규성 검정

```python
# data 불러오기
with open("./datas2/성인여성_키_데이터.txt", "r") as f:
    data = f.read().split('\n')
    # print(data)
    # print(list(map(float, data)))
    data = list(map(float, data))
    print(data)
    print(len(data))
```

## 데이터의 평균과 표준편차 구하기

```python
from scipy import stats
import numpy as np

# 평균과 표준편차 계산
mean_value = np.mean(data)
std_dev = np.std(data, ddof=1)  # 모집단 표준편차(기본 설정)

mean_value, std_dev
```

## 가설 설정
- 귀무가설: 표본 데이터의 평균은 163과 같다.
- 대립가설: 표본 데이터의 평균은 163과 다르다.

## 정규성 검정
- 수집한 표본데이터가 정규분포를 따르는지 검증한다.
### 정규성 검정의 가설
- 귀무가설 : 수집한 표본의 데이터는 정규성을 따른다.
- 대립가설 : 수집한 표본의 데이터는 정규성을 따르지 않는다.

```python
# 정규성 검정, KS-test(Kolmogorov-Smirnov) 검정
#  KS-test는 주어진 표본 데이터가 특정 이론적인 분포(예: 정규분포)를 따르는지 검정하는 방법

# 'norm': 비교 대상이 되는 이론적인 분포를 지정 (stats.norm.cdf은 정규분포)
# args=(np.mean(data), np.std(data, ddof=0)): 표본 데이터의 평균과 표준편차를 사용하여 정규분포의 모수 설정
ks_statistic, p_value = stats.kstest(data, stats.norm.cdf, args=(np.mean(data), np.std(data, ddof=0)))
ks_statistic, p_value

# p-value > 0.05 : 정규분포를 따르지 않는다라고 말할 충분한 증거가 없음. 귀무가설 기각하지 않음
```

## [문제]
- 귀무가설: 표본 데이터의 평균은 163과 같다.
- 대립가설: 표본 데이터의 평균은 163과 다르다.

```python
# 단일 표본 t 검정 수행
# 검중 수치 = 163, 귀무가설 평균키는 163이다.

print(stats.ttest_1samp(data, 163)) 
```

### 결과 해석
- p-value < 0.05 이므로 귀무가설 기각/대립가설 채택 -> 표본 데이터의 평균은 163과 다르다

- 결과적으로 여성 키의 표본 데이터의 평균은 163과 다르다는 것이 통계적으로 유의미함

- 이후에 큰지 작은지는 또 귀무가설과 대립가설 세워서 더 검정 가능
